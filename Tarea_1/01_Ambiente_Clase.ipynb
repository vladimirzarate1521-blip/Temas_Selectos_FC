# 4. Creación del ambiente

A cotinuacion se muestra el paso a paso de la creacion del ambiente de trabajo y capturas que muestran algunas salidas de la terminal, cabe resaltar que todo fue ejecutado desde PowerShel, sin embargo el proceso tambien puede ser llevado por medio de Anaconda Prompt

## 4.1 Crear el ambiente

Ejecuta en la terminal:

```bash
conda create -n fisica-computacional python=3.10
```

Esta instrucción crea un nuevo ambiente Conda llamado `fisica-computacional` e instala **Python 3.10** dentro de él.

Cuando aparezca:

```text
Proceed ([y]/n)?
```

Conda está solicitando confirmación para continuar con la instalación. Escribe:

```text
y
```

y presiona `Enter`.

Al finalizar, el ambiente estará creado. 

---

## 4.2 Activar el ambiente

Ejecuta en la terminal:

```bash
conda activate fisica-computacional
```

Esta instrucción **activa el ambiente Conda** que creamos anteriormente.


Después de activarlo, el nombre del ambiente deberá aparecer al inicio del prompt:

```text
(fisica-computacional)
```

Esto indica que los comandos `python`, `pip` y las bibliotecas que instalemos se utilizarán dentro de este ambiente.

A continuacion se muestra evidencia de la existencia del ambiente de trabajo activacion del mismo:

## 4.3 Verificar Python

Ejecuta en la terminal:

```bash
python --version
```

Esta instrucción muestra la **versión de Python** que está utilizando actualmente la terminal.

Deberás obtener algo semejante a:

```text
Python 3.10.x
```

La `x` representa una versión de mantenimiento de Python 3.10, por ejemplo `3.10.12`.

## 15. Prueba con partículas

Crea `prueba_particula.py`:

```python
from particle import Particle

electron = Particle.from_pdgid(11)
muon = Particle.from_pdgid(13)
top = Particle.from_pdgid(6)

print(electron)
print(muon)
print(top)
```

`Particle.from_pdgid(...)` identifica una partícula mediante su **PDG ID**:

```text
11 → electrón
13 → muón
 6 → quark top
```

Ejecuta:

```bash
python prueba_particula.py
```

## 16. Prueba con cuadrivectores

Crea `prueba_vector.py`:

```python
import vector

p = vector.obj(
    px=30.0,
    py=40.0,
    pz=100.0,
    E=120.0,
)

print("pT =", p.pt)
print("|p| =", p.p)
print("masa invariante =", p.mass)
```

`vector.obj(...)` construye un cuadrivector a partir de las componentes del momento y la energía.

```text
p.pt   → momento transversal
p.p    → magnitud del trimomento
p.mass → masa invariante
```

Ejecuta:

```bash
python prueba_vector.py
```
